```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
# Provinces in Indonesia list
provinces = ["Aceh", "Riau", "Bengkulu", "Sumatera Selatan", "Sumatera Barat", "Sumatera Utara", "Jambi", "Lampung"]
```

```python
# Get csv files with company information

# Change directory to access all csv files
cwd = os.getcwd() # Get current working directory
list_cwd = os.listdir() # List all files and directories inside the present one
os.chdir(cwd + "\\" + list_cwd[1]) # Change directory to the one with the csv's

# Get all csvs
companies_list = os.listdir()
companies_list
```

```python
# Dictionary for company dataframe information
companies = {}

# Save the company name and the dataframe
for company in companies_list:

    name = company.split("-")[0] # Get the first word of the file name
    companies[name] = pd.read_csv(company) # Save csv as dataframe

# Example to visualize what's in the dictionary  
companies
```

```python
# Clean the dataframes and only save the rows with Sumatra provinces
def prepare_df(df):

  """
  Create a dataframe with only information of Sumatra, Indonesia
  """

  # Save data from Sumatra in specific (cut the rest of the columns and rows)
  df = df[df["Province"].isin(provinces)]

  # Save the columns we consider important
  df = df.loc[:, "UML ID":"sum_of_treeloss_km"].reset_index()
  df.drop(columns = df.loc[:, "Confidence level" : "treeloss_ha_2022"], inplace = True)

  # Sort data by province and parent company columns
  df_prov = df.loc[:, ["Province", "Parent Company", "RSPO Status", "sum_of_treeloss_km"]].sort_values(by = ["Province", "Parent Company"])
  df_prov = df_prov.groupby(["Province", "Parent Company", "RSPO Status"])["sum_of_treeloss_km"].sum()

  return df_prov.reset_index()
```

```python
# Prepare the dataframe for each company
for company in companies:

    company_df = companies[company] # Dataframe
    companies[company] = prepare_df(company_df) # Edit dataframe

companies
```

```python
# Dictionary with mill information
mill_info = pd.DataFrame()

# Iterate through the company dataframes and provinces
for company in companies:
  for p in provinces:
    
    # Temporary df with information of a specific province
    specific_company = companies[company]
    temp = specific_company[specific_company["Province"] == p]
    mill_count = temp["Parent Company"].nunique()

    # Group by the province and companies and add the sum of treeloss
    temp = temp.groupby(["Province"])["sum_of_treeloss_km"].sum().reset_index()
    temp["Mill Count"] = mill_count
    temp["Company"] = company
    
    # Save values in a dataframe
    mill_info = pd.concat([mill_info, temp])

# Reorganize the columns in the dataframe
mill_info = mill_info[["Company", "Province", "Mill Count", "sum_of_treeloss_km"]].reset_index().drop(columns = ["index"])
mill_info.head(24)
```
